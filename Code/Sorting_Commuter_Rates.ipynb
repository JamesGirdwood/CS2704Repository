```python
# Creates a new table based on Rate_allmodes from the Commuter Rates sheet
# Returns the Area Name and Rate_allmodes, sorting the table by Rate_allmodes
import pandas as pd

# Load the spreadsheet
file_path = "Active Commuter Rates.xlsx"  # Replace with the actual file path
sheet_name = "2020_ACS"  # Specify the sheet name

# Read the spreadsheet into a DataFrame
df = pd.read_excel(file_path, sheet_name=sheet_name)

# Select the 'Geographic Area Name' and 'Rate_allmodes' columns
selected_columns = df[['Geographic Area Name', 'Rate_allmodes']]

# Sort the DataFrame by 'Rate_allmodes' in descending order
sorted_df = selected_columns.sort_values(by='Rate_allmodes', ascending=False)

# Reset the index for clean output
sorted_df.reset_index(drop=True, inplace=True)

# Display the sorted DataFrame
print(sorted_df)

# Optional: Save the sorted DataFrame to a new Excel file
sorted_df.to_excel("Sorted_Rate_allmodes.xlsx", index=False)
```
